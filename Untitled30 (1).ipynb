```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


matches = pd.read_csv(r"C:\Users\cgc\Downloads\matches (1).csv")
deliveries = pd.read_csv(r"C:\Users\cgc\Downloads\deliveries (2).csv")

df = pd.merge(deliveries,
              matches,
              left_on='match_id',
              right_on='id')

print("Shape =", df.shape)

print("\nColumn Names")
print(df.columns)

print("\nData Types")
print(df.dtypes)

print("\nSeasons Covered =", matches['season'].nunique())
```

```python
print(matches.isnull().sum())

matches['city'] = matches['city'].fillna(
    matches['city'].mode()[0]
)

matches['player_of_match'] = matches['player_of_match'].fillna(
    "No Award"
)

deliveries['dismissal_kind'] = deliveries['dismissal_kind'].fillna(
    "not out"
)
```

```python
df['total_runs'] = df['batsman_runs'] + df['extra_runs']

df['is_boundary'] = np.where(
    df['batsman_runs'] >= 4,
    'boundary',
    'non-boundary'
)

def phase(over):
    if over <= 6:
        return "Powerplay"
    elif over <= 15:
        return "Middle"
    else:
        return "Death"

df['over_phase'] = df['over'].apply(phase)


print(df[['batsman_runs',
          'extra_runs',
          'total_runs',
          'is_boundary',
          'over',
          'over_phase']].head(10))
```

```python
top5 = df.groupby('batsman')['batsman_runs'].sum()

top5 = top5.sort_values(ascending=False).head()

print(top5)
```

```python
balls = deliveries.groupby('bowler').size()

runs = deliveries.groupby('bowler')['total_runs'].sum()

economy = runs / (balls / 6)

economy = economy[balls >= 500]

print(economy)
```

```python
matches['win'] = 1

season_win = matches.groupby(
    ['season', 'winner']
)['win'].count().reset_index()

season_total = matches.groupby(
    'season'
)['win'].count().reset_index()

season_total.rename(
    columns={'win': 'total_matches'},
    inplace=True
)

season_win = season_win.merge(
    season_total,
    on='season'
)

season_win['win_percentage'] = (
    season_win['win']
    / season_win['total_matches']
) * 100

result = season_win.loc[
    season_win.groupby(
        'season'
    )['win_percentage'].idxmax()
]

print(result)
```

```python
toss_data = matches.groupby('toss_decision')['winner'].count()

toss_percent = (toss_data / toss_data.sum()) * 100

print(toss_percent)

plt.figure(figsize=(6,5))

sns.barplot(x=toss_percent.index,
            y=toss_percent.values,
            hue=toss_percent.index)

plt.title("Percentage of Matches by Toss Decision")
plt.xlabel("Toss Decision")
plt.ylabel("Percentage (%)")

plt.show()
```

```python
top10 = df.groupby('batsman')['batsman_runs'].sum()

top10 = top10.sort_values(ascending=False).head(10)

print(top10)

plt.figure(figsize=(8,6))

top10.sort_values().plot(kind='barh')

plt.title("Top 10 Run Scorers")
plt.xlabel("Runs")
plt.ylabel("Batsman")

plt.show()
```

```python
top3 = top10.head(3).index

dismiss = deliveries[
    deliveries['batsman'].isin(top3)
].copy()

dismiss['dismissal_kind'] = dismiss['dismissal_kind'].fillna('not out')

pivot = pd.crosstab(
    dismiss['batsman'],
    dismiss['dismissal_kind']
)

print(pivot)

pivot.plot(
    kind='bar',
    stacked=True,
    figsize=(8,6)
)

plt.title("Dismissal Pattern of Top 3 Batsmen")
plt.xlabel("Batsman")
plt.ylabel("Count")

plt.show()
```

```python
season_runs = df.groupby('season')['total_runs'].sum()

season_matches = matches.groupby('season')['id'].count()

avg_runs = season_runs / season_matches

print(avg_runs)

plt.figure(figsize=(8,5))

plt.plot(avg_runs.index,
         avg_runs.values,
         marker='o')

plt.title("Average Runs Per Match by Season")
plt.xlabel("Season")
plt.ylabel("Average Runs")

plt.grid()

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.regplot(x=avg_runs.index,
            y=avg_runs.values)

plt.title("Season vs Average Runs")

plt.show()
```

```python
top5_venues = matches['venue'].value_counts().head(5)

print(top5_venues)
```

```python
venue_df = matches[
    matches['venue'].isin(top5_venues.index)
]

heat = pd.crosstab(
    venue_df['venue'],
    venue_df['winner']
)

print(heat)

plt.figure(figsize=(12,6))

sns.heatmap(heat,
            cmap='Blues')

plt.title("Win Count by Venue")

plt.show()
```

```python
death = deliveries[
    (deliveries['over'] >= 16) &
    (deliveries['over'] <= 20)
]

print(death.head())
```

```python
top6 = death.groupby(
    'batting_team'
)['total_runs'].sum()

top6 = top6.sort_values(
    ascending=False
).head(6)

print(top6)
```

```python
death_team = death[
    death['batting_team'].isin(top6.index)
]

plt.figure(figsize=(10,6))

sns.boxplot(data=death_team,
            x='batting_team',
            y='total_runs')

plt.xticks(rotation=90)

plt.title("Death Over Performance")

plt.xlabel("Batting Team")
plt.ylabel("Runs")

plt.show()
```

```python
1. Executive Summary

The IPL dataset analysis reveals important patterns in batting performance, match outcomes, and team strategies. Over the years, the average runs per match have increased, indicating that the IPL has become more batting-friendly. Toss decisions, venue conditions, and death-over performance significantly influence the result of matches. Top batsmen consistently contribute large scores, while some teams show exceptional performance during the final overs.
```

```python
2. Key Insights
Insight 1: Top Batsmen Dominate Run Scoring
Players like Virat Kohli, Suresh Raina, and Rohit Sharma are among the highest run scorers.
Their consistency over multiple seasons makes them valuable assets for their teams.

Interpretation:
The batting performance of a few star players contributes significantly to team success.

Insight 2: IPL is Becoming More High-Scoring
Average runs per match show an increasing trend across seasons.
The regression trend line indicates a positive relationship between season and average runs.

Interpretation:
Modern T20 cricket has become more aggressive, with teams focusing on higher strike rates and power hitting.

Insight 3: Toss Decision Alone Does Not Guarantee Victory
Most teams prefer to field first after winning the toss.
However, the difference in winning percentages between batting first and fielding first is not very large.

Interpretation:
Team performance and match conditions have a greater impact than toss decisions.

Insight 4: Venue Conditions Affect Match Results
Certain teams perform exceptionally well at particular stadiums.
Some venues provide a noticeable home advantage.

Interpretation:
Familiar pitch conditions and crowd support help teams achieve better results.

Insight 5: Death Overs Play a Crucial Role
Teams with strong finishers score heavily during overs 16–20.
Some teams show both high median scores and high variability in death overs.

Interpretation:
Effective death-over batting can significantly change the outcome of a match.

3. Recommendation for Team Strategy

Based on the analysis, teams should focus on strengthening their middle-order and death-over batting. Captains should prefer chasing targets when conditions are favorable, but decisions should depend more on pitch and weather conditions rather than relying solely on the toss. Selecting players according to venue characteristics can also improve the chances of winning.

4. Self-Reflection

One limitation of the dataset is the absence of detailed player statistics such as strike rate under pressure, pitch conditions, weather information, and field placements. These factors have a strong influence on match outcomes but are not available in the dataset.

If additional data were available, deeper analysis could be performed on player matchups, venue-specific strategies, batting partnerships, and the impact of weather and pitch conditions on team performance.

Conclusion

The analysis demonstrates that batting strength, death-over performance, and venue conditions are key factors influencing IPL results. While toss decisions have some impact, overall team performance and strategic planning play a much more important role in determining success in the IPL.
```
